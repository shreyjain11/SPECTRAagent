## 1. Data source/acquisition

Read the supplied investigation record and verify its content fingerprint. The record identifies which candidate axes were actually evaluated.


```python
from pathlib import Path
import json,hashlib,socket,os
import pandas as pd
from IPython.display import display
record_path=Path.cwd()/'inputs/investigation_record.md'
payload=record_path.read_bytes()
assert hashlib.sha256(payload).hexdigest()=='3e83e4ccd4052f695e59da095a81d3eb532159af254de47b4de614242410107c','Investigation record fingerprint differs'
record=payload.decode()
```

## 2. Data processing

Extract the candidate-disposition row for membrane topology.


```python
rows=[]
for line in record.splitlines():
    fields=[value.strip() for value in line.strip().strip('|').split('|')]
    if len(fields)==4 and fields[0]=='A06':rows.append(dict(zip(['candidate_id','axis','disposition','reason'],fields)))
if len(rows)!=1:raise ValueError('Expected exactly one membrane-topology disposition')
candidate=rows[0]
display(pd.DataFrame(rows))
```
